## A Bell state

A Hadamard on qubit 0 followed by a CNOT gives $\frac{1}{\sqrt{2}}(|00\rangle + |11\rangle)$: measuring both qubits should give `00` or `11`, each about half the time, and never `01` or `10`.

```python
from qiskit import QuantumCircuit

bell = QuantumCircuit(2, 2)
bell.h(0)
bell.cx(0, 1)
bell.measure([0, 1], [0, 1])

bell.draw("mpl")
```

## Run it on the Aer simulator

```python
from qiskit import transpile
from qiskit_aer import AerSimulator

simulator = AerSimulator()
job = simulator.run(transpile(bell, simulator), shots=1024, seed_simulator=42)
counts = job.result().get_counts()
print(counts)

assert set(counts) <= {"00", "11"}, f"unexpected outcomes: {counts}"
assert sum(counts.values()) == 1024
```

```python
from qiskit.visualization import plot_histogram

plot_histogram(counts, title="Bell state, 1024 shots")
```
